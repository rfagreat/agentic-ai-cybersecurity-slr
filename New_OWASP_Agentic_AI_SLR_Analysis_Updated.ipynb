## 3. OWASP Agentic AI Top 10 — Threat Scores Dataset

```python
owasp_data = {
    'ID':               ['ASI01','ASI02','ASI03','ASI04','ASI05','ASI06','ASI07','ASI08','ASI09','ASI10'],
    'Threat_Category':  ['Agent Goal Hijack','Tool Misuse & Exploitation','Identity & Privilege Abuse',
                         'Unexpected Code Execution','Insecure Inter-Agent Comm.',
                         'Human-Agent Trust Exploit.','Supply Chain Vulnerabilities',
                         'Memory & Context Poisoning','Cascading Failures','Rogue Agents'],
    'Short_Name':       ['Goal Hijack','Tool Misuse','Identity Abuse','Code Exec',
                         'Insecure Comm','Human Trust','Supply Chain','Memory Poison',
                         'Cascading','Rogue Agents'],
    'Severity':         [9.5, 8.8, 9.2, 8.5, 7.8, 7.5, 8.0, 8.3, 8.7, 9.0],
    'Likelihood':       [9.0, 8.5, 8.8, 7.5, 7.0, 8.0, 7.5, 7.8, 7.2, 6.5],
    'Attack_Complexity': [3.0, 5.0, 4.0, 6.5, 6.0, 3.5, 5.5, 5.0, 7.0, 7.5],
    'Potential_Impact':  [9.5, 8.5, 9.0, 8.5, 7.5, 7.0, 8.0, 8.0, 9.0, 9.5],
    'Num_Studies':       [28, 12, 10, 9, 10, 9, 11, 11, 11, 11],
}

df_owasp = pd.DataFrame(owasp_data)
df_owasp['Risk_Score'] = (df_owasp['Severity'] * df_owasp['Likelihood'] / 10).round(2)
df_owasp['Severity_Likelihood_Gap'] = (df_owasp['Severity'] - df_owasp['Likelihood']).round(1)
df_owasp['Exploitability'] = (10 - df_owasp['Attack_Complexity']).round(1)
df_owasp['Composite_Risk'] = ((df_owasp['Severity'] * 0.3 + df_owasp['Likelihood'] * 0.3 +
                                df_owasp['Potential_Impact'] * 0.25 + df_owasp['Exploitability'] * 0.15)).round(2)

def classify_risk(score):
    if score >= 8.0: return 'Critical'
    elif score >= 7.0: return 'High'
    elif score >= 6.0: return 'Medium'
    else: return 'Low'

df_owasp['Risk_Level'] = df_owasp['Risk_Score'].apply(classify_risk)
print('✅ OWASP ASI Top 10 Scores:')
df_owasp[['ID','Threat_Category','Severity','Likelihood','Risk_Score',
           'Attack_Complexity','Potential_Impact','Num_Studies','Risk_Level']]
```

## 4. Study-to-Threat Mapping (Which Papers Support Each ASI Category)

```python
study_to_threat = {
    'ASI01': ['S001','S002','S003','S004','S005','S006','S007','S008','S010',
              'S011','S016','S019','S020','S021','S022','S025','S029','S030',
              'S031','S033','S038','S039','S046','S049','S051','S053','S063','S066'],
    'ASI02': ['S009','S025','S026','S033','S034','S037','S042','S053','S056',
              'S060','S068','S072'],
    'ASI03': ['S015','S026','S034','S044','S048','S054','S058','S062','S064','S070'],
    'ASI04': ['S004','S029','S033','S040','S043','S056','S060','S062','S068'],
    'ASI05': ['S012','S019','S034','S035','S052','S053','S058','S062','S065','S075'],
    'ASI06': ['S006','S014','S017','S018','S035','S045','S050','S067','S071'],
    'ASI07': ['S023','S024','S034','S037','S048','S055','S059','S063','S067',
              'S072','S075'],
    'ASI08': ['S019','S023','S024','S038','S051','S053','S054','S061','S063',
              'S066','S073'],
    'ASI09': ['S019','S026','S034','S052','S055','S058','S062','S065','S067',
              'S074','S075'],
    'ASI10': ['S017','S026','S040','S047','S050','S052','S057','S064','S069',
              'S072','S073'],
}

all_study_ids = df_studies['ID'].tolist()
threat_ids = list(study_to_threat.keys())
mapping_matrix = pd.DataFrame(0, index=all_study_ids, columns=threat_ids)
for threat, studies in study_to_threat.items():
    for study in studies:
        if study in mapping_matrix.index:
            mapping_matrix.loc[study, threat] = 1

mapping_matrix['Year'] = df_studies.set_index('ID')['Year']
mapping_matrix['Category'] = df_studies.set_index('ID')['Category']
mapping_matrix['Threats_Covered'] = mapping_matrix[threat_ids].sum(axis=1)

print('✅ Study-to-Threat Mapping Matrix created')
print(f'   Total mappings: {sum(len(v) for v in study_to_threat.values())}')
print(f'   Studies mapped to ≥1 threat: {(mapping_matrix["Threats_Covered"]>0).sum()}')
print(f'   Studies mapped to ≥2 threats: {(mapping_matrix["Threats_Covered"]>1).sum()}')
print(f'\nTop multi-threat studies:')
multi = mapping_matrix[mapping_matrix['Threats_Covered'] > 1][['Category','Threats_Covered']].sort_values('Threats_Covered', ascending=False)
print(multi.head(15))
```

## 19. 📊 Qualitative Classification Decision Flowchart

```python
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(14, 16), dpi=150)
ax.set_xlim(0, 14); ax.set_ylim(0, 16); ax.axis('off')
ax.set_facecolor('#fafafa')

fc = {'start':'#2C3E50','decision':'#F7DC6F','process':'#85C1E9',
      'ATK':'#E74C3C','DEF':'#27AE60','SUR':'#3498DB','THR':'#F39C12','FRM':'#9B59B6'}

def draw_box(ax, x, y, w, h, text, color, shape='rect', fs=9, fc2='black'):
    if shape == 'diamond':
        ax.add_patch(plt.Polygon([(x,y+h/2),(x+w/2,y),(x,y-h/2),(x-w/2,y)],
                     facecolor=color, edgecolor='#333', lw=1.5))
    elif shape == 'oval':
        ax.add_patch(mpatches.Ellipse((x,y), w, h, facecolor=color, edgecolor='white', lw=2))
        fc2 = 'white'
    else:
        ax.add_patch(FancyBboxPatch((x-w/2,y-h/2), w, h,
                     boxstyle='round,pad=0.02,rounding_size=0.2',
                     facecolor=color, edgecolor='#333', lw=1.5))
    ax.text(x, y, text, ha='center', va='center', fontsize=fs, color=fc2,
            weight='normal', multialignment='center')

def darr(ax, s, e, txt='', to=(0,0)):
    ax.annotate('', xy=e, xytext=s, arrowprops=dict(arrowstyle='->', color='#333', lw=1.5))
    if txt:
        mx, my = (s[0]+e[0])/2+to[0], (s[1]+e[1])/2+to[1]
        ax.text(mx, my, txt, fontsize=8, ha='center', va='center', color='#333', weight='bold',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='none'))

draw_box(ax, 7, 15.3, 2.5, 0.6, 'Read Full Paper', fc['start'], 'oval', 10)
draw_box(ax, 7, 14.2, 4, 0.7, 'Step 1: Identify Primary\nResearch Focus', fc['process'], fs=9)
darr(ax, (7,14.95), (7,14.55))

draw_box(ax, 7, 12.8, 3.5, 1.2, 'Is it an industry\nframework, guideline,\nor standard?', fc['decision'], 'diamond', 8)
darr(ax, (7,13.85), (7,13.4))
draw_box(ax, 12, 12.8, 2, 0.8, 'FRM\n(n=9)', fc['FRM'], fs=10, fc2='white')
darr(ax, (8.75,12.8), (11,12.8), 'Yes')

draw_box(ax, 7, 11, 3.5, 1.2, 'Is it primarily a\nsurvey, review, or\nfoundational work?', fc['decision'], 'diamond', 8)
darr(ax, (7,12.2), (7,11.6), 'No')
draw_box(ax, 12, 11, 2, 0.8, 'SUR\n(n=18)', fc['SUR'], fs=10, fc2='white')
darr(ax, (8.75,11), (11,11), 'Yes')

draw_box(ax, 7, 9.2, 3.5, 1.2, 'Does it demonstrate\nor develop new\nattack techniques?', fc['decision'], 'diamond', 8)
darr(ax, (7,10.4), (7,9.8), 'No')
draw_box(ax, 12, 9.2, 2, 0.8, 'ATK\n(n=23)', fc['ATK'], fs=10, fc2='white')
darr(ax, (8.75,9.2), (11,9.2), 'Yes')

draw_box(ax, 7, 7.4, 3.5, 1.2, 'Does it propose\ndefense mechanisms\nor security tools?', fc['decision'], 'diamond', 8)
darr(ax, (7,8.6), (7,8), 'No')
draw_box(ax, 12, 7.4, 2, 0.8, 'DEF\n(n=14)', fc['DEF'], fs=10, fc2='white')
darr(ax, (8.75,7.4), (11,7.4), 'Yes')

draw_box(ax, 7, 5.6, 3.5, 1.2, 'Does it analyze\nthreats, vulnerabilities,\nor risk assessment?', fc['decision'], 'diamond', 8)
darr(ax, (7,6.8), (7,6.2), 'No')
draw_box(ax, 12, 5.6, 2, 0.8, 'THR\n(n=11)', fc['THR'], fs=10, fc2='white')
darr(ax, (8.75,5.6), (11,5.6), 'Yes')

draw_box(ax, 7, 4, 3.5, 0.7, 'Re-evaluate: Check\nsecondary focus', fc['process'], fs=9)
darr(ax, (7,5), (7,4.35), 'No')
ax.annotate('', xy=(5.25,12.8), xytext=(5.25,4),
            arrowprops=dict(arrowstyle='->', color='#7f8c8d', lw=1.5, connectionstyle='angle,angleA=0,angleB=90'))
ax.plot([5.25,5.25], [4,4.35], color='#7f8c8d', lw=1.5)

cx = 2
for y, txt in [(12.8,'• OWASP, NIST, CSA, IEEE standards\n• Industry guidelines\n• Governance documents'),
               (11,'• Literature reviews\n• Agent architectures\n• Tool learning surveys'),
               (9.2,'• Prompt injection attacks\n• Jailbreaking methods\n• Autonomous exploitation'),
               (7.4,'• Detection systems\n• Guardrails & filters\n• Sandboxing & privilege control'),
               (5.6,'• RAG poisoning analysis\n• Worm propagation study\n• Vulnerability & risk assessment')]:
    ax.text(cx, y, txt, fontsize=7, va='center', ha='left', color='#666',
            bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))

summary = 'CLASSIFICATION SUMMARY (n=75)\n━━━━━━━━━━━━━━━━━━━━━━━━\n'
summary += 'ATK: Attack Research        23 (30.7%)\n'
summary += 'SUR: Surveys & Foundations   18 (24.0%)\n'
summary += 'DEF: Defensive Applications  14 (18.7%)\n'
summary += 'THR: Threat Analysis         11 (14.7%)\n'
summary += 'FRM: Industry Frameworks      9 (12.0%)'
ax.text(7, 2.2, summary, fontsize=9, va='center', ha='center', family='monospace',
        bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#bdc3c7', lw=2))

legend_elements = [mpatches.Patch(facecolor=fc[k], label=l) for k, l in
    [('start','Start/End'),('decision','Decision'),('process','Process'),
     ('ATK','ATK: Attack'),('DEF','DEF: Defense'),('SUR','SUR: Survey'),
     ('THR','THR: Threat'),('FRM','FRM: Framework')]]
ax.legend(handles=legend_elements, loc='lower right', fontsize=8, framealpha=0.95, fancybox=True, ncol=2)

ax.text(7, 16.2, 'Fig. X: Qualitative Classification Decision Flowchart',
        ha='center', fontsize=14, weight='bold', color=fc['start'])
ax.text(7, 15.8, 'Decision process for categorizing studies in the systematic literature review',
        ha='center', fontsize=10, style='italic', color='gray')

plt.tight_layout()
plt.savefig('fig_classification_flowchart.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print('✅ Classification Flowchart saved')
```
