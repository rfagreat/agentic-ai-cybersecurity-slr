## 3. OWASP Agentic AI Top 10 — Threat Scores Dataset

```python
# ============================================================
# OWASP ASI Top 10: Literature-derived scores (UPDATED)
# Num_Studies updated to match verified study-to-threat mapping
# ============================================================

owasp_data = {
    'ID':               ['ASI01','ASI02','ASI03','ASI04','ASI05','ASI06','ASI07','ASI08','ASI09','ASI10'],
    'Threat_Category':  ['Agent Goal Hijack','Tool Misuse & Exploitation','Identity & Privilege Abuse',
                         'Unexpected Code Execution','Insecure Inter-Agent Comm.',
                         'Human-Agent Trust Exploit.','Supply Chain Vulnerabilities',
                         'Memory & Context Poisoning','Cascading Failures','Rogue Agents'],
    'Short_Name':       ['Goal Hijack','Tool Misuse','Identity Abuse','Code Exec',
                         'Insecure Comm','Human Trust','Supply Chain','Memory Poison',
                         'Cascading','Rogue Agents'],
    'Severity':         [9.5, 8.8, 9.2, 8.5, 7.8, 7.5, 8.0, 8.3, 8.7, 9.0],
    'Likelihood':       [9.0, 8.5, 8.8, 7.5, 7.0, 8.0, 7.5, 7.8, 7.2, 6.5],
    'Attack_Complexity': [3.0, 5.0, 4.0, 6.5, 6.0, 3.5, 5.5, 5.0, 7.0, 7.5],
    'Potential_Impact':  [9.5, 8.5, 9.0, 8.5, 7.5, 7.0, 8.0, 8.0, 9.0, 9.5],
    'Num_Studies':       [28, 12, 10, 9, 10, 9, 11, 11, 11, 11],
}

df_owasp = pd.DataFrame(owasp_data)

# Derived columns
df_owasp['Risk_Score'] = (df_owasp['Severity'] * df_owasp['Likelihood'] / 10).round(2)
df_owasp['Severity_Likelihood_Gap'] = (df_owasp['Severity'] - df_owasp['Likelihood']).round(1)
df_owasp['Exploitability'] = (10 - df_owasp['Attack_Complexity']).round(1)
df_owasp['Composite_Risk'] = ((df_owasp['Severity'] * 0.3 + df_owasp['Likelihood'] * 0.3 +
                                df_owasp['Potential_Impact'] * 0.25 + df_owasp['Exploitability'] * 0.15)).round(2)

def classify_risk(score):
    if score >= 8.0: return 'Critical'
    elif score >= 7.0: return 'High'
    elif score >= 6.0: return 'Medium'
    else: return 'Low'

df_owasp['Risk_Level'] = df_owasp['Risk_Score'].apply(classify_risk)

print('✅ OWASP ASI Top 10 Scores (Updated):')
df_owasp[['ID','Threat_Category','Severity','Likelihood','Risk_Score',
           'Attack_Complexity','Potential_Impact','Num_Studies','Risk_Level']]
```

## 4. Study-to-Threat Mapping (Which Papers Support Each ASI Category)

```python
# ============================================================
# MAPPING: Each study → OWASP ASI categories it supports
# UPDATED for 75 verified studies (Feb 2026)
# A study can map to multiple threat categories
# ============================================================

study_to_threat = {
    # ASI01: Agent Goal Hijack — prompt injection, jailbreaking, goal manipulation
    'ASI01': ['S001','S002','S003','S004','S005','S006','S007','S008','S010',
              'S011','S016','S019','S020','S021','S022','S025','S029','S030',
              'S031','S033','S038','S039','S046','S049','S051','S053','S063','S066'],

    # ASI02: Tool Misuse & Exploitation — tool-calling vulnerabilities, agent tool abuse
    'ASI02': ['S009','S025','S026','S033','S034','S037','S042','S053','S056',
              'S060','S068','S072'],

    # ASI03: Identity & Privilege Abuse — privilege escalation, identity spoofing in agents
    'ASI03': ['S015','S026','S034','S044','S048','S054','S058','S062','S064','S070'],

    # ASI04: Unexpected Code Execution — code injection via LLMs, neural exec
    'ASI04': ['S004','S029','S033','S040','S043','S056','S060','S062','S068'],

    # ASI05: Insecure Inter-Agent Communication — multi-agent trust, propagation
    'ASI05': ['S012','S019','S034','S035','S052','S053','S058','S062','S065','S075'],

    # ASI06: Human-Agent Trust Exploitation — social engineering via AI, misplaced trust
    'ASI06': ['S006','S014','S017','S018','S035','S045','S050','S067','S071'],

    # ASI07: Supply Chain Vulnerabilities — poisoned models, compromised tools/plugins
    'ASI07': ['S023','S024','S034','S037','S048','S055','S059','S063','S067',
              'S072','S075'],

    # ASI08: Memory & Context Poisoning — RAG attacks, memory manipulation, context injection
    'ASI08': ['S019','S023','S024','S038','S051','S053','S054','S061','S063',
              'S066','S073'],

    # ASI09: Cascading Failures — multi-agent failure propagation, systemic risks
    'ASI09': ['S019','S026','S034','S052','S055','S058','S062','S065','S067',
              'S074','S075'],

    # ASI10: Rogue Agents — autonomous misalignment, uncontrolled agent behaviour
    'ASI10': ['S017','S026','S040','S047','S050','S052','S057','S064','S069',
              'S072','S073'],
}

# Create binary mapping matrix (studies × threats)
all_study_ids = df_studies['ID'].tolist()
threat_ids = list(study_to_threat.keys())

mapping_matrix = pd.DataFrame(0, index=all_study_ids, columns=threat_ids)
for threat, studies in study_to_threat.items():
    for study in studies:
        if study in mapping_matrix.index:
            mapping_matrix.loc[study, threat] = 1

mapping_matrix['Year'] = df_studies.set_index('ID')['Year']
mapping_matrix['Category'] = df_studies.set_index('ID')['Category']
mapping_matrix['Threats_Covered'] = mapping_matrix[threat_ids].sum(axis=1)

print('✅ Study-to-Threat Mapping Matrix created')
print(f'\nStudies covering multiple threats:')
multi = mapping_matrix[mapping_matrix['Threats_Covered'] > 1][['Category','Threats_Covered']].sort_values('Threats_Covered', ascending=False)
print(multi.head(15))
print(f'\nStudies not mapped to any ASI category: {(mapping_matrix["Threats_Covered"]==0).sum()}')
print(f'Total study-threat mappings: {sum(len(v) for v in study_to_threat.values())}')
```

## 19. 📊 Fig. X — Qualitative Classification Decision Flowchart

Decision process for categorizing each of the 75 studies into ATK, DEF, SUR, THR, or FRM.

```python
"""
Qualitative Classification Decision Flowchart
for "When Agentic AI Meets Cybersecurity" SLR (75 studies)
"""

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(14, 16), dpi=150)
ax.set_xlim(0, 14)
ax.set_ylim(0, 16)
ax.axis('off')
ax.set_facecolor('#fafafa')

colors = {
    'start': '#2C3E50', 'decision': '#F7DC6F', 'process': '#85C1E9',
    'ATK': '#E74C3C', 'DEF': '#27AE60', 'SUR': '#3498DB',
    'THR': '#F39C12', 'FRM': '#9B59B6',
}

def draw_box(ax, x, y, width, height, text, color, shape='rect', fontsize=9, fontcolor='black'):
    if shape == 'diamond':
        diamond = plt.Polygon([(x, y+height/2), (x+width/2, y), (x, y-height/2), (x-width/2, y)],
                              facecolor=color, edgecolor='#333', linewidth=1.5)
        ax.add_patch(diamond)
    elif shape == 'oval':
        ellipse = mpatches.Ellipse((x, y), width, height, facecolor=color,
                                    edgecolor='white', linewidth=2)
        ax.add_patch(ellipse)
        fontcolor = 'white'
    else:
        box = FancyBboxPatch((x - width/2, y - height/2), width, height,
                             boxstyle="round,pad=0.02,rounding_size=0.2",
                             facecolor=color, edgecolor='#333', linewidth=1.5)
        ax.add_patch(box)
    ax.text(x, y, text, ha='center', va='center', fontsize=fontsize,
            color=fontcolor, weight='normal', wrap=True, multialignment='center')
    return (x, y)

def draw_arrow(ax, start, end, text='', text_offset=(0, 0)):
    ax.annotate('', xy=end, xytext=start,
                arrowprops=dict(arrowstyle='->', color='#333', lw=1.5))
    if text:
        mid_x = (start[0] + end[0]) / 2 + text_offset[0]
        mid_y = (start[1] + end[1]) / 2 + text_offset[1]
        ax.text(mid_x, mid_y, text, fontsize=8, ha='center', va='center',
                color='#333', weight='bold',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='none'))

# FLOWCHART STRUCTURE
start = draw_box(ax, 7, 15.3, 2.5, 0.6, 'Read Full Paper', colors['start'], 'oval', fontsize=10)

step1 = draw_box(ax, 7, 14.2, 4, 0.7, 'Step 1: Identify Primary\nResearch Focus', colors['process'], fontsize=9)
draw_arrow(ax, (7, 14.95), (7, 14.55))

# Decision 1: Framework?
dec1 = draw_box(ax, 7, 12.8, 3.5, 1.2, 'Is it an industry\nframework, guideline,\nor standard?',
                colors['decision'], 'diamond', fontsize=8)
draw_arrow(ax, (7, 13.85), (7, 13.4))

frm_box = draw_box(ax, 12, 12.8, 2, 0.8, 'FRM\n(n=9)', colors['FRM'], fontsize=10, fontcolor='white')
draw_arrow(ax, (8.75, 12.8), (11, 12.8), 'Yes')

# Decision 2: Survey?
dec2 = draw_box(ax, 7, 11, 3.5, 1.2, 'Is it primarily a\nsurvey, review, or\nfoundational work?',
                colors['decision'], 'diamond', fontsize=8)
draw_arrow(ax, (7, 12.2), (7, 11.6), 'No')

sur_box = draw_box(ax, 12, 11, 2, 0.8, 'SUR\n(n=18)', colors['SUR'], fontsize=10, fontcolor='white')
draw_arrow(ax, (8.75, 11), (11, 11), 'Yes')

# Decision 3: Attack?
dec3 = draw_box(ax, 7, 9.2, 3.5, 1.2, 'Does it demonstrate\nor develop new\nattack techniques?',
                colors['decision'], 'diamond', fontsize=8)
draw_arrow(ax, (7, 10.4), (7, 9.8), 'No')

atk_box = draw_box(ax, 12, 9.2, 2, 0.8, 'ATK\n(n=23)', colors['ATK'], fontsize=10, fontcolor='white')
draw_arrow(ax, (8.75, 9.2), (11, 9.2), 'Yes')

# Decision 4: Defense?
dec4 = draw_box(ax, 7, 7.4, 3.5, 1.2, 'Does it propose\ndefense mechanisms\nor security tools?',
                colors['decision'], 'diamond', fontsize=8)
draw_arrow(ax, (7, 8.6), (7, 8), 'No')

def_box = draw_box(ax, 12, 7.4, 2, 0.8, 'DEF\n(n=14)', colors['DEF'], fontsize=10, fontcolor='white')
draw_arrow(ax, (8.75, 7.4), (11, 7.4), 'Yes')

# Decision 5: Threat?
dec5 = draw_box(ax, 7, 5.6, 3.5, 1.2, 'Does it analyze\nthreats, vulnerabilities,\nor risk assessment?',
                colors['decision'], 'diamond', fontsize=8)
draw_arrow(ax, (7, 6.8), (7, 6.2), 'No')

thr_box = draw_box(ax, 12, 5.6, 2, 0.8, 'THR\n(n=11)', colors['THR'], fontsize=10, fontcolor='white')
draw_arrow(ax, (8.75, 5.6), (11, 5.6), 'Yes')

# Re-evaluate loop
reeval = draw_box(ax, 7, 4, 3.5, 0.7, 'Re-evaluate: Check\nsecondary focus', colors['process'], fontsize=9)
draw_arrow(ax, (7, 5), (7, 4.35), 'No')

ax.annotate('', xy=(5.25, 12.8), xytext=(5.25, 4),
            arrowprops=dict(arrowstyle='->', color='#7f8c8d', lw=1.5,
                           connectionstyle='angle,angleA=0,angleB=90'))
ax.plot([5.25, 5.25], [4, 4.35], color='#7f8c8d', lw=1.5)

# Criteria boxes (left side)
criteria_x = 2
ax.text(criteria_x, 12.8, '• OWASP, NIST, CSA, IEEE standards\n• Industry guidelines\n• Governance documents',
        fontsize=7, va='center', ha='left', color='#666',
        bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))
ax.text(criteria_x, 11, '• Literature reviews\n• Agent architectures\n• Tool learning surveys',
        fontsize=7, va='center', ha='left', color='#666',
        bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))
ax.text(criteria_x, 9.2, '• Prompt injection attacks\n• Jailbreaking methods\n• Autonomous exploitation',
        fontsize=7, va='center', ha='left', color='#666',
        bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))
ax.text(criteria_x, 7.4, '• Detection systems\n• Guardrails & filters\n• Sandboxing & privilege control',
        fontsize=7, va='center', ha='left', color='#666',
        bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))
ax.text(criteria_x, 5.6, '• RAG poisoning analysis\n• Worm propagation study\n• Vulnerability & risk assessment',
        fontsize=7, va='center', ha='left', color='#666',
        bbox=dict(boxstyle='round', facecolor='#f8f9fa', edgecolor='#ddd'))

# Summary box
summary_text = """CLASSIFICATION SUMMARY (n=75)
━━━━━━━━━━━━━━━━━━━━━━━━
ATK: Attack Research        23 (30.7%)
SUR: Surveys & Foundations   18 (24.0%)
DEF: Defensive Applications  14 (18.7%)
THR: Threat Analysis         11 (14.7%)
FRM: Industry Frameworks      9 (12.0%)"""

ax.text(7, 2.2, summary_text, fontsize=9, va='center', ha='center',
        family='monospace',
        bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#bdc3c7', linewidth=2))

# Legend
legend_elements = [
    mpatches.Patch(facecolor=colors['start'], label='Start/End'),
    mpatches.Patch(facecolor=colors['decision'], label='Decision'),
    mpatches.Patch(facecolor=colors['process'], label='Process'),
    mpatches.Patch(facecolor=colors['ATK'], label='ATK: Attack'),
    mpatches.Patch(facecolor=colors['DEF'], label='DEF: Defense'),
    mpatches.Patch(facecolor=colors['SUR'], label='SUR: Survey'),
    mpatches.Patch(facecolor=colors['THR'], label='THR: Threat'),
    mpatches.Patch(facecolor=colors['FRM'], label='FRM: Framework'),
]
ax.legend(handles=legend_elements, loc='lower right', fontsize=8,
          framealpha=0.95, fancybox=True, ncol=2)

# Title
ax.text(7, 16.2, 'Fig. X: Qualitative Classification Decision Flowchart',
        ha='center', fontsize=14, weight='bold', color=colors['start'])
ax.text(7, 15.8, 'Decision process for categorizing studies in the systematic literature review',
        ha='center', fontsize=10, style='italic', color='gray')

plt.tight_layout()
plt.savefig('fig_classification_flowchart.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
print('✅ Classification Flowchart saved')
```
